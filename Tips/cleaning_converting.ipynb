```python
# =====================================================================
#                    GENERIC DATA CLEANING PIPELINE
# =====================================================================
#
# OBJECTIF
# ---------------------------------------------------------------------
# Préparer un dataset brut pour l'analyse en :
#
#   1. Comprenant les données
#   2. Supprimant les doublons
#   3. Traitant les valeurs manquantes
#   4. Convertissant les types de données
#   5. Vérifiant la qualité du résultat final
#
#
# POURQUOI ?
# ---------------------------------------------------------------------
# Les données brutes contiennent souvent :
#
# - des doublons
# - des valeurs manquantes
# - des types incorrects
# - des textes mal formatés
#
# Si ces problèmes ne sont pas corrigés :
#
# - les statistiques deviennent fausses
# - les visualisations deviennent trompeuses
# - les modèles de Machine Learning sont dégradés
#
#
# WORKFLOW GLOBAL
# ---------------------------------------------------------------------
#
#                    RAW DATASET
#
#                         │
#                         ▼
#
#                 1. INSPECTION
#
#                         │
#                         ▼
#
#                 2. DUPLICATES
#
#                         │
#                         ▼
#
#              3. MISSING VALUES
#
#                         │
#                         ▼
#
#             4. TYPE CONVERSION
#
#                         │
#                         ▼
#
#               5. VERIFICATION
#
#                         │
#                         ▼
#
#                  CLEAN DATASET
#
#                         │
#                         ▼
#
#             EDA / VISUALIZATION
#
#                         │
#                         ▼
#
#               STATISTICS / ML
#
#
# =====================================================================
# ETAPE 1 : INSPECTION
# =====================================================================
#
# OBJECTIF
# ---------------------------------------------------------------------
# Comprendre les données avant toute modification.
#
# Questions auxquelles répondre :
#
# - Combien de lignes ?
# - Combien de colonnes ?
# - Quels sont les types ?
# - Y a-t-il des valeurs manquantes ?
# - Quelle mémoire est utilisée ?
#
#
# Fonctions :
#
# inspect_dataframe()
#
#
# Résultat :
#
# Une vue globale du dataset.
#
#
# =====================================================================
# ETAPE 2 : DUPLICATES
# =====================================================================
#
# OBJECTIF
# ---------------------------------------------------------------------
# Supprimer les observations répétées.
#
# Pourquoi ?
#
# Les doublons peuvent :
#
# - gonfler artificiellement certaines catégories
# - fausser les moyennes
# - fausser les distributions
# - biaiser les modèles prédictifs
#
#
# Workflow :
#
# inspect_duplicates()
#         │
#         ▼
# show_duplicates()
#         │
#         ▼
# remove_duplicates()
#         │
#         ▼
# handle_duplicates()
#
#
# Exemple :
#
# client_id    achat
#
# 1001         500
# 1001         500
#
# Le même achat apparaît deux fois.
#
#
# Résultat :
#
# Chaque observation n'existe qu'une seule fois.
#
#
# =====================================================================
# ETAPE 3 : MISSING VALUES
# =====================================================================
#
# OBJECTIF
# ---------------------------------------------------------------------
# Gérer les données manquantes de manière cohérente.
#
#
# STRATEGIE UTILISEE
# ---------------------------------------------------------------------
#
# Missing % > 50%
#
#       ▼
#
# Drop Column
#
#
# Pourquoi ?
#
# Une colonne avec plus de la moitié des données
# manquantes contient trop peu d'information.
#
#
# --------------------------------------------------------------
#
# 5% < Missing % <= 50%
#
#       ▼
#
# Imputation
#
#
# Numérique :
#
#       Median
#
#
# Pourquoi la médiane ?
#
# Elle est robuste aux valeurs extrêmes.
#
#
# Exemple :
#
# 20, 25, 30, 5000
#
# Moyenne :
#
# 1268.75
#
# Médiane :
#
# 27.5
#
# La médiane représente mieux la tendance centrale.
#
#
# --------------------------------------------------------------
#
# Catégorielle :
#
#       Most Frequent (Mode)
#
#
# Pourquoi ?
#
# On remplace la valeur manquante
# par la catégorie la plus représentée.
#
#
# Exemple :
#
# Homme
# Homme
# Femme
# NaN
#
# devient
#
# Homme
# Homme
# Femme
# Homme
#
#
# --------------------------------------------------------------
#
# Missing % <= 5%
#
#       ▼
#
# Drop Rows
#
#
# Pourquoi ?
#
# La perte de données est faible.
#
# Il est souvent plus simple
# de supprimer ces observations.
#
#
# Workflow :
#
# inspect_missing_values()
#          │
#          ▼
# drop_high_missing_columns()
#          │
#          ▼
# drop_low_missing_rows()
#          │
#          ▼
# impute_medium_missing_columns()
#          │
#          ▼
# handle_missing_values()
#
#
# Résultat :
#
# Dataset sans valeurs manquantes.
#
#
# =====================================================================
# ETAPE 4 : TYPE CONVERSION
# =====================================================================
#
# OBJECTIF
# ---------------------------------------------------------------------
# Convertir les colonnes vers le type le plus adapté.
#
#
# Pourquoi ?
#
# Les datasets réels contiennent souvent :
#
# "25"
# "2024-01-01"
# "Paris"
#
# stockés comme texte.
#
#
# Or :
#
# 25 doit être numérique
#
# 2024-01-01 doit être datetime
#
# Paris peut devenir category
#
#
# TYPES CIBLES
# ---------------------------------------------------------------------
#
# Numeric
#
# Utilisation :
#
# - statistiques
# - corrélations
# - machine learning
#
#
# Datetime
#
# Utilisation :
#
# - séries temporelles
# - regroupement par année/mois
#
#
# Category
#
# Utilisation :
#
# - optimisation mémoire
# - variables répétitives
#
#
# String
#
# Utilisation :
#
# - identifiants
# - texte libre
#
#
# Workflow :
#
# detect_numeric_columns()
# detect_date_columns()
# detect_category_columns()
#
#           ▼
#
# Validation humaine
#
#           ▼
#
# convert_numeric_columns()
# convert_date_columns()
# convert_category_columns()
#
#
# ATTENTION
# ---------------------------------------------------------------------
#
# Certaines colonnes peuvent être mal détectées.
#
# Exemple :
#
# zipcode
#
# 75001
# 75002
# 75003
#
# semble numérique
#
# MAIS
#
# représente un identifiant.
#
#
# Une validation humaine reste donc nécessaire.
#
#
# =====================================================================
# ETAPE 5 : VERIFICATION
# =====================================================================
#
# OBJECTIF
# ---------------------------------------------------------------------
# Vérifier que le nettoyage s'est bien déroulé.
#
#
# Contrôles :
#
# - plus de doublons
# - plus de NaN
# - types corrects
# - mémoire optimisée
#
#
# Fonctions :
#
# inspect_dataframe()
#
#
# Résultat :
#
# Dataset prêt pour l'analyse.
#
#
# =====================================================================
# PIPELINE PRINCIPAL
# =====================================================================
#
# clean_dataset(df)
#
#         │
#         ▼
#
# inspect_dataframe()
#
#         │
#         ▼
#
# handle_duplicates()
#
#         │
#         ▼
#
# handle_missing_values()
#
#         │
#         ▼
#
# auto_clean_dataframe()
#
#         │
#         ▼
#
# inspect_dataframe()
#
#         │
#         ▼
#
# CLEAN DATASET
#
#
# =====================================================================
# APRES LE NETTOYAGE
# =====================================================================
#
# df.describe()
#
# df.groupby(...)
#
# sns.histplot(...)
#
# sns.boxplot(...)
#
# plt.plot(...)
#
# ttest_ind(...)
#
# shapiro(...)
#
# train_test_split(...)
#
#
# =====================================================================
# PHILOSOPHIE DU PIPELINE
# =====================================================================
#
# Comprendre
#      ↓
# Nettoyer
#      ↓
# Vérifier
#      ↓
# Analyser
#      ↓
# Interpréter
#
# =====================================================================
```
