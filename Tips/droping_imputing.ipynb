# Suppression colonnes / imputation

### Cas 1 : Tout imputer après suppression des colonnes > 50% 

```python

from sklearn.impute import SimpleImputer

df = []  # Assuming df is your DataFrame after loading the dataset


# ==========================================
# 1. Analyse des valeurs manquantes
# ==========================================

missing_percentage = (df.isnull().sum() / len(df)) * 100

# ==========================================
# 2. Suppression des colonnes > 50%
# ==========================================

drop_cols = missing_percentage[ missing_percentage > 50].index

df = df.drop(columns=drop_cols)

# ==========================================
# 3. Séparation des variables
# ==========================================

num_cols = df.select_dtypes(include='number').columns
cat_cols = df.select_dtypes(include=['object', 'category']).columns

# ==========================================
# 4. Imputation
# ==========================================

num_imputer = SimpleImputer(strategy='median')
cat_imputer = SimpleImputer(strategy='most_frequent')

df[num_cols] = num_imputer.fit_transform(df[num_cols])
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])

# ==========================================
# 5. Vérification
# ==========================================

print("Valeurs manquantes restantes :", df.isnull().sum().sum())
```

### Cas 2 : Supprimer les lignes pour les colonnes < 5%, puis imputer le reste

```python
from sklearn.impute import SimpleImputer

# ==========================================
# 1. Analyse des valeurs manquantes
# ==========================================

missing_percentage = (df.isnull().sum() / len(df)) * 100

# ==========================================
# 2. Suppression des colonnes > 50%
# ==========================================

drop_cols = missing_percentage[
    missing_percentage > 50
].index

df = df.drop(columns=drop_cols)

# ==========================================
# 3. Recalcul des pourcentages
# ==========================================

missing_percentage = (df.isnull().sum() / len(df)) * 100

# ==========================================
# 4. Colonnes entre 0 et 5%
# ==========================================

low_missing_cols = missing_percentage[
    (missing_percentage > 0) &
    (missing_percentage < 5)
].index

print("Colonnes concernées :", list(low_missing_cols))

# ==========================================
# 5. Suppression des lignes
# ==========================================

df = df.dropna(subset=low_missing_cols)

# ==========================================
# 6. Séparation des variables
# ==========================================

num_cols = df.select_dtypes(include='number').columns
cat_cols = df.select_dtypes(include=['object', 'category']).columns

# ==========================================
# 7. Imputation du reste
# ==========================================

num_imputer = SimpleImputer(strategy='median')
cat_imputer = SimpleImputer(strategy='most_frequent')

df[num_cols] = num_imputer.fit_transform(df[num_cols])
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])

# ==========================================
# 8. Vérification
# ==========================================

print("Valeurs manquantes restantes :", df.isnull().sum().sum())
```
