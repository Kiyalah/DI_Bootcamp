## Exercise 2 - Line Plot

```python

temperatures = [72, 74, 76, 80, 82, 78, 75]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

plt.figure(figsize=(8,4))
plt.plot(days, temperatures, marker="o")
plt.title("Temperature Variation Over a Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.show()
```

## Exercise 3 - Bar Chart

```python

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(8,4))
plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.show()
```

## Exercise 4 - CGPA Distribution

```python
df = pd.read_csv("data/Student Mental health.csv")

print(df.columns.tolist())
```

```python

plt.figure(figsize=(8,4))
sns.histplot(df["What is your CGPA?"], color="skyblue")
plt.title("Distribution of CGPA")
plt.xlabel("CGPA Range")
plt.show()
```

## Exercise 5 - Anxiety Across Gender

```python

anxiety_gender = pd.crosstab(
    df["Choose your gender"],
    df["Do you have Anxiety?"],
    normalize="index"
)

anxiety_gender.plot(kind="bar", figsize=(8,4))
plt.title("Anxiety Levels Across Genders")
plt.ylabel("Proportion")
plt.show()
```

## Exercise 6 - Age vs Panic Attacks

```python

df["Panic_Attack_Numeric"] = df["Do you have Panic attack?"].map({
    "Yes": 1,
    "No": 0
})

plt.figure(figsize=(8,4))
sns.scatterplot(
    data=df,
    x="Age",
    y="Panic_Attack_Numeric"
)

plt.title("Relationship Between Age and Panic Attacks")
plt.xlabel("Age")
plt.ylabel("Panic Attack (1=Yes, 0=No)")
plt.show()
```
