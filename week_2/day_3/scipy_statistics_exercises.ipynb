```python
import scipy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats
from scipy.stats import norm, ttest_ind, linregress, f_oneway, binom, pearsonr, spearmanr
```

# 🌟 Exercise 2: Descriptive Statistics

```python

data = [12, 15, 13, 12, 18, 20, 22, 21]

# Using scipy.stats and numpy for median
print("Mean:", stats.tmean(data))
print("Median:", np.median(data))
print("Variance:", stats.tvar(data))
print("Standard Deviation:", stats.tstd(data))
```

# 🌟 Exercise 3: Understanding Distributions

```python

data = np.linspace(20, 80, 100)

mean = 50
std_dev = 10

y = norm.pdf(data, mean, std_dev)

plt.figure(figsize=(8,5))
plt.plot(data, y)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Probability Density")

plt.show()

# plt.hist(data, bins=30)

```

# 🌟 Exercise 4: T-Test Application

```python

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = ttest_ind(data1, data2)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

# 🌟 Exercise 5: Linear Regression Analysis

```python

house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

slope, intercept, r_value, p_value, std_err = linregress(house_sizes, house_prices)

print("Slope:", slope)
print("Intercept:", intercept)

predicted_price = slope * 90 + intercept

print("Predicted Price for 90m²:", predicted_price)

print("\nInterpretation:")
print("The slope represents the increase in house price for each additional square meter.")
```

# 🌟 Exercise 6: Understanding ANOVA

```python

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print("F-Value:", f_value)
print("P-Value:", p_value)

if p_value < 0.05:
    print("\nThe fertilizers have significantly different effects.")
    print("We reject the null hypothesis.")
else:
    print("\nThe fertilizers do not have significantly different effects.")
    print("We fail to reject the null hypothesis.")

```

# 🌟 Exercise 7: Probability Distributions

```python

probability = binom.pmf(5, 10, 0.5)

print("Probability of getting exactly 5 heads:", probability)
```

# 🌟 Exercise 8: Correlation Coefficients

```python

data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

pearson_corr, _ = pearsonr(data['age'], data['income'])
spearman_corr, _ = spearmanr(data['age'], data['income'])

print("Pearson Correlation:", pearson_corr)
print("Spearman Correlation:", spearman_corr)
```
