```python
# ==============================================================================
# 1. INSTALLATION DES BIBLIOTHÈQUES
# ==============================================================================
%pip install kaggle pandas

import os
import shutil
import pandas as pd

print("=== ÉTAPE 1 : CONFIGURATION DES IDENTIFIANTS KAGGLE ===")

# Définition des chemins pour la clé API sur macOS
source_json = os.path.expanduser("~/Downloads/kaggle.json")
destination_json = os.path.expanduser("~/.kaggle/kaggle.json")

# Création du dossier masqué .kaggle s'il n'existe pas
os.makedirs(os.path.dirname(destination_json), exist_ok=True)

# Déplacement et sécurisation de la clé d'authentification
if os.path.exists(source_json):
    shutil.move(source_json, destination_json)
    os.chmod(destination_json, 0o600)
    print("Le fichier kaggle.json a été déplacé et sécurisé avec succès !")
elif os.path.exists(destination_json):
    print("Le fichier kaggle.json est déjà en place et prêt à l'emploi.")
else:
    print("⚠️ ATTENTION : Placez le fichier 'kaggle.json' téléchargé du site dans votre dossier Téléchargements.")

# ==============================================================================
# 2. TÉLÉCHARGEMENT ET EXTRACTION DU JEU DE DONNÉES TITANIC
# ==============================================================================
print("\n=== ÉTAPE 2 : TÉLÉCHARGEMENT DU DATASET ===")

# Définition dynamique du dossier de stockage du projet
data_dir = os.path.join(os.getcwd(), "titanic_data")
zip_path = os.path.join(data_dir, "titanic.zip")

# Téléchargement uniquement si les données ne sont pas déjà là
if not os.path.exists(os.path.join(data_dir, "train.csv")):
    print(f"Téléchargement en cours dans : {data_dir}...")
    
    # Télécharge l'archive de la compétition via le CLI Kaggle
    !kaggle competitions download -c titanic -p "{data_dir}"
    
    # Extraction de l'archive ZIP sur Mac
    !unzip -o "{zip_path}" -d "{data_dir}"
    
    # Nettoyage du fichier ZIP d'origine
    if os.path.exists(zip_path):
        os.remove(zip_path)
    print("Téléchargement et décompression terminés !")
else:
    print("Les fichiers CSV du Titanic sont déjà présents dans votre dossier de travail.")

```
