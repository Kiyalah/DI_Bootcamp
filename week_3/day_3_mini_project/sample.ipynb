```python
df = stock_prices_dataset.copy()
df
```

```python
# Examining the time series properties of the data
# Dates are in day/month/year format, so parse with dayfirst=True
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
```

```python
# stock_prices_dataset.set_index("Date", inplace=True)
# stock_prices_dataset.sort_index(inplace=True)
# stock_prices_dataset.plot(y="Close", title="Apple Stock Closing Prices Over Time")


# ...existing code...
# updated plotting cell: use Date as index and show only the year on the x-axis
df.set_index("Date", inplace=True)
df.sort_index(inplace=True)


ax = df["Close"].plot(title="Apple Stock Closing Prices Over Time", figsize=(10, 4))
ax.set_xlabel("Date")
ax.set_ylabel("Closing Price (USD)")

# afficher seulement l'année
ax.xaxis.set_major_locator(mdates.YearLocator(1))        # 1 tick par an (mettre 5 pour un tick tous les 5 ans)
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
# ...existing code...
```
