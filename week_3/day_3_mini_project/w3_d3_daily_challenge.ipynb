### Transformation des données catégorielles en format numérique

En regardant les informations de notre tableau (`df.info()`), on remarque que toutes les colonnes sont déjà des nombres (`int64` ou `float64`).

Cependant, certains nombres représentent en fait des **catégories** :
- **Choix binaires (Oui/Non) :** `blue` (Bluetooth), `dual_sim`, `four_g`, `three_g`, `touch_screen`, `wifi`. (0 = Non, 1 = Oui).
- **Catégories nominales :** `n_cores` (Nombre de cœurs, de 1 à 8).

Pour que l'ordinateur comprenne mieux ces catégories, on utilise une technique appelée **"One-Hot Encoding"**. Cela crée de nouvelles colonnes pour chaque possibilité.

Par exemple, pour le Bluetooth, au lieu d'une colonne avec 0 ou 1, on aura une colonne qui confirme la présence de l'option.

```python
# On crée une copie de nos données pour ne pas modifier l'original
df_final = df.copy()

# 1. Transformation pour le Bluetooth ('blue')
# 'drop_first=True' évite d'avoir des colonnes redondantes (si ce n'est pas 0, c'est forcément 1)
df_final = pd.get_dummies(df_final, columns=['blue'], prefix='bluetooth', drop_first=True)

# 2. Transformation pour le nombre de cœurs ('n_cores')
# Cela va créer 8 colonnes (n_cores_1, n_cores_2, etc.)
df_final = pd.get_dummies(df_final, columns=['n_cores'], prefix='coeurs')

# Affichons un aperçu des nouvelles colonnes créées
print("Nouvelles colonnes après transformation :")
display(df_final.head())

print(f"\nAncienne forme : {df.shape}")
print(f"Nouvelle forme : {df_final.shape}")
```

### Résumé des modifications

- La colonne `blue` est devenue `bluetooth_1`.
- La colonne `n_cores` a été divisée en plusieurs colonnes `coeurs_1`, `coeurs_2`, etc.
- La valeur **1** dans ces nouvelles colonnes signifie que l'appareil appartient à cette catégorie.

C'est une étape essentielle pour préparer les données avant de les donner à un algorithme d'apprentissage automatique (Machine Learning) !

### 3. Analyse Statistique avec NumPy et SciPy

Nous allons approfondir l'étude des caractéristiques numériques pour comprendre leur distribution et leur importance par rapport à la variable cible `price_range`.

```python
import numpy as np
from scipy import stats

# Sélection des colonnes numériques principales pour l'analyse
features = ['battery_power', 'clock_speed', 'ram', 'mobile_wt', 'int_memory']

stats_dict = {
    'Caractéristique': [],
    'Moyenne': [],
    'Médiane': [],
    'Écart-type': [],
    'Skewness (Asymétrie)': [],
    'Kurtosis (Aplatissement)': []
}

for col in features:
    data = df[col]
    stats_dict['Caractéristique'].append(col)
    stats_dict['Moyenne'].append(np.mean(data))
    stats_dict['Médiane'].append(np.median(data))
    stats_dict['Écart-type'].append(np.std(data))
    stats_dict['Skewness (Asymétrie)'].append(stats.skew(data))
    stats_dict['Kurtosis (Aplatissement)'].append(stats.kurtosis(data))

df_stats = pd.DataFrame(stats_dict)
display(df_stats)
```

#### Étude des corrélations

Voyons quelles caractéristiques influencent le plus le prix du téléphone.

```python
# Calcul de la corrélation de Spearman (adaptée aux rangs de prix)
correlations = {}
for col in df.columns:
    if col != 'price_range':
        coef, p_val = stats.spearmanr(df[col], df['price_range'])
        correlations[col] = coef

# Tri par importance absolue
sorted_corr = sorted(correlations.items(), key=lambda x: abs(x[1]), reverse=True)

print("Top 5 des caractéristiques les plus corrélées au prix :")
for feature, coef in sorted_corr[:5]:
    print(f"{feature}: {coef:.4f}")
```

### 4. Visualisation des données avec Matplotlib et Seaborn

La visualisation nous aide à valider les statistiques calculées précédemment. Nous allons nous concentrer sur la distribution du prix et sa relation avec la RAM.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration du style des graphiques
sns.set_theme(style="whitegrid")
plt.figure(figsize=(15, 5))

# 1. Distribution de la variable cible (Price Range)
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='price_range', palette='viridis', hue='price_range', legend=False)
plt.title('Répartition des catégories de prix')
plt.xlabel('Gamme de prix (0=Bas, 3=Très Haut)')
plt.ylabel('Nombre de téléphones')

# 2. Relation entre la RAM et le Prix
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='price_range', y='ram', palette='magma', hue='price_range', legend=False)
plt.title('Impact de la RAM sur la gamme de prix')
plt.xlabel('Gamme de prix')
plt.ylabel('RAM (Mo)')

plt.tight_layout()
plt.show()
```

#### Carte de chaleur (Heatmap) des corrélations

Ce graphique permet de voir d'un coup d'œil quelles variables évoluent ensemble.

```python
plt.figure(figsize=(12, 10))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Carte de chaleur des corrélations entre toutes les caractéristiques')
plt.show()
```
