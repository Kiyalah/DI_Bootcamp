```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
import scipy.stats as stats
```

```python
initial_dataset = pd.read_csv("data/global_power_plant_database.csv")
initial_dataset.head()
```

```python
initial_dataset.describe()
```

```python
df = initial_dataset.copy()
```

```python
# Checking for duplicates and dropping them
count_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {count_duplicates}")
df = df.drop_duplicates()
print(f"Number of duplicate rows after dropping: {df.duplicated().sum()}")
```

```python
count_missing_values = df.isnull().sum()
print(f"Number of missing values in each column:\n{count_missing_values}")
```

```python
# Handling missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing %': missing_percentage
})

missing_df.sort_values('Missing %', ascending=False)
```

```python
missing_df_2 = (df.isnull().mean() * 100)\
    .sort_values(ascending=False)\
    .to_frame('Missing %')
    
missing_df_2
```

```python
# 1. Colonnes > 50%
drop_cols = missing_percentage[missing_percentage > 50].index

df = df.drop(columns=drop_cols)

# 2. Colonnes numériques
num_cols = df.select_dtypes(include='number').columns

# 3. Colonnes catégorielles
cat_cols = df.select_dtypes(include=['object', 'str']).columns
```

```python
num_imputer = SimpleImputer(strategy='median')
cat_imputer = SimpleImputer(strategy='most_frequent')

df[num_cols] = num_imputer.fit_transform(df[num_cols])
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
```

```python
print("Checking for Missing Values:\n")
print(df.isnull().sum())
```

```python
df.info()
```

```python
# Safely convert relevant columns to numerical types if necessary.
# If a column named 'numeric_column' exists, convert it.
# Otherwise, fall back to converting the numeric columns identified earlier (num_cols).

if 'numeric_column' in df.columns:
	cols_to_convert = ['numeric_column']
else:
	# num_cols is defined in an earlier cell and contains the numeric columns we want to ensure are numeric
	cols_to_convert = list(num_cols)

# Record missing counts before conversion
before_na = df[cols_to_convert].isna().sum()

# Convert columns to numeric (coerce errors to NaN)
for col in cols_to_convert:
	df[col] = pd.to_numeric(df[col], errors='coerce')

# Record missing counts after conversion
after_na = df[cols_to_convert].isna().sum()

print("Missing values before conversion:\n", before_na)
print("Missing values after conversion:\n", after_na)
print("Newly introduced missing values (after - before):\n", (after_na - before_na))
```

```python
# Record missing counts before conversion
before_na = df[cols_to_convert].isna().sum()

# Use numpy.vectorize to safely coerce values to float (non-convertible -> np.nan)
def _safe_float(x):
    try:
        return float(x)
    except Exception:
        return np.nan

_vec = np.vectorize(_safe_float)

for col in cols_to_convert:
    df[col] = _vec(df[col].values)

# Record missing counts after conversion
after_na = df[cols_to_convert].isna().sum()

print("Missing values before conversion:\n", before_na)
print("Missing values after conversion:\n", after_na)
print("Newly introduced missing values (after - before):\n", (after_na - before_na))
```

```python
# Statistical Analysis:
# Perform a statistical analysis of power output by fuel type using NumPy’s statistical functions.
# Use hypothesis testing to determine if the mean power output differs significantly between different fuel types.


# Choose a numeric column that represents power output. Prefer capacity_mw, fallback to estimated_generation_gwh_2017
if 'capacity_mw' in df_cleaned.columns:
	value_col = 'capacity_mw'
elif 'estimated_generation_gwh_2017' in df_cleaned.columns:
	value_col = 'estimated_generation_gwh_2017'
else:
	raise KeyError("No suitable numeric column found for power output analysis (expected 'capacity_mw' or 'estimated_generation_gwh_2017').")

# Group the data by primary fuel type and calculate the mean power output for each group
mean_power_by_fuel = df_cleaned.groupby('primary_fuel')[value_col].mean()
print(f"Mean {value_col} by fuel:\n", mean_power_by_fuel)

# Build groups of numeric arrays for ANOVA (drop NaNs and require at least 2 observations per group)
groups = [
	group[value_col].dropna().values
	for _, group in df_cleaned.groupby('primary_fuel')
	if group[value_col].dropna().size >= 2
]

if len(groups) < 2:
	print("Not enough groups with numeric data to perform ANOVA.")
else:
	# Perform a one-way ANOVA test
	f_stat, p_value = stats.f_oneway(*groups)
	print("\nANOVA F-statistic:", f_stat)
	print("ANOVA p-value:", p_value)
```

```python
# Time Series Analysis:

# If the dataset includes time-related data (like year of establishment), use NumPy to analyze trends over time.
# Explore how the mix of fuel types for power generation has evolved over the years.

# Choose the best available year column and analyze trends by count and capacity
candidate_year_cols = ['year_of_establishment', 'commissioning_year', 'year_of_capacity_data']
available = [c for c in candidate_year_cols if c in df_cleaned.columns]

if not available:
    print("No suitable year column found. Cannot perform time series analysis.")
else:
    # pick column with the most non-null values
    year_col = max(available, key=lambda c: df_cleaned[c].notna().sum())
    non_null_count = df_cleaned[year_col].notna().sum()
    if non_null_count == 0:
        print(f"Year column {year_col!r} has no non-null values. Cannot perform time series analysis.")
    else:
        # Coerce to numeric years
        df_cleaned[year_col] = pd.to_numeric(df_cleaned[year_col], errors='coerce').astype('Int64')

        # Drop rows missing year or primary_fuel
        tmp = df_cleaned.dropna(subset=[year_col, 'primary_fuel'])

        # Count of plants per year and fuel
        fuel_trends_count = tmp.groupby([year_col, 'primary_fuel']).size().unstack(fill_value=0).sort_index()

        # Sum of capacity (MW) per year and fuel
        if 'capacity_mw' in tmp.columns:
            fuel_trends_capacity = tmp.groupby([year_col, 'primary_fuel'])['capacity_mw'].sum().unstack(fill_value=0).sort_index()
        else:
            fuel_trends_capacity = None

        print(f"Using year column: {year_col!r} (non-null: {non_null_count})\n")
        print("Plant counts by year and fuel (first 10 years):")
        print(fuel_trends_count.head(10))
        if fuel_trends_capacity is not None:
            print("\nTotal capacity (MW) by year and fuel (first 10 years):")
            print(fuel_trends_capacity.head(10))
```
