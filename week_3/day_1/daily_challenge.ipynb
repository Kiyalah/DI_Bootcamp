```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
temp_data = np.random.uniform(low=-5, high=35, size=(10, 5))
print("Temperature Data (°C):")
print(temp_data)
```

```python
temp_df = pd.DataFrame(temp_data, columns=[f"Day_{i+1}" for i in range(temp_data.shape[1])])
print("\nTemperature DataFrame:")
print(temp_df)
```

```python
cities = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix", "Philadelphia", "San Antonio", "San Diego", "Dallas", "San Jose"]
temp_df = temp_df.set_index(pd.Index(cities))
print("Temperature DataFrame with City Names: \n")
print(temp_df)
```

```python
temp_avg_per_city = temp_df.mean(axis=1)
print("Average Temperature for Each City:\n")
print(temp_avg_per_city)
```

```python
highest_city_temp = temp_avg_per_city.idxmax()
print("\nCity with the Highest Average Temperature:")
print(highest_city_temp)
```

```python
lowest_city_temp = temp_avg_per_city.idxmin()
print("\nCity with the Lowest Average Temperature:")
print(lowest_city_temp)
```

```python
new_temp_data = np.random.uniform(low=-5, high=35, size=(10, 12))
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
new_temp_df = pd.DataFrame(new_temp_data, columns=months).set_index(pd.Index(cities))
new_temp_df
```

```python
# With Matplotlib
plt.figure(figsize=(14,8))

new_temp_df.T.plot(ax=plt.gca())

plt.title("Monthly Temperature Trends for Each City")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.grid(True)

plt.show()
```

```python
# With Seaborn

plt.figure(figsize=(14,8))

for city in new_temp_df.index:
    sns.lineplot(
        x=new_temp_df.columns,
        y=new_temp_df.loc[city],
        label=city
    )

plt.title("Monthly Temperature Trends for Each City")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)
plt.show()
```
