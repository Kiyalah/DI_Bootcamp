```python
import numpy as np
```

```python
array_1d = np.arange(10)
print(array_1d)
```

```python
array_1d[2, ]
```

```python
list = [3.14, 2.17, 0, 1, 2]
converted_list = np.array(list)
converted_list = converted_list.astype(int)
print(converted_list)
```

```python
array_2d = np.arange(1,10).reshape(3,3)
print(array_2d)
```

```python
array_2d[1, :]
```

```python
arr_2d = np.random.rand(4,5)
print(arr_2d)
```

```python
array = np.array([[21,22,23,22,22],[20, 21, 22, 23, 24],[21,22,23,22,22]])
second_row = array[1, :]
print(f"The second row of the array is: {second_row}")
```

```python
arr_6 = np.arange(1, 10)
print(f"The original array is: {arr_6}")
arr_reversed = arr_6[::-1]
print(f"The reversed array is: {arr_reversed}")
```

```python
id_matrix = np.eye(4)
print(id_matrix)
```

```python
arr_8 = np.arange(1,16)
print(f"The original array is: {arr_8}")
arr_8_sum = arr_8.sum()
print(f"The sum of the array is: {arr_8_sum}")
arr_8_mean = arr_8.mean()
print(f"The mean of the array is: {arr_8_mean}")
```

```python
array_9 = np.arange(1,21)
reshaped_array_9 = array_9.reshape(4,5)
print(f"The reshaped array is:\n{reshaped_array_9}")
```

```python
array_10 = np.arange(1, 10)
result = array_10[array_10 % 2 != 0]
print(f"The odd numbers in the array are: {result}")
```
