## 🌟 Exercise 2: LLM Architecture and Application Scenarios

### BERT
- **Architecture**: Bidirectional encoder-only; Masked Language Modeling (MLM).
- **Application**: Named Entity Recognition (NER) or Sentiment Analysis.
- **Why**: BERT looks at the context both to the left and to the right of a word simultaneously, making it excellent at understanding the precise meaning of words within a fixed sentence.

### GPT
- **Architecture**: Decoder-only; autoregressive causal LM.
- **Application**: Creative writing assistants or Chatbots.
- **Why**: GPT is designed to predict the next token in a sequence, which makes it naturally suited for generating long, coherent, and creative strings of text.

### T5
- **Architecture**: Encoder-decoder; text-to-text framework.
- **Application**: Document Summarization or Translation.
- **Why**: T5 uses a flexible framework where every task is treated as converting input text to output text, allowing it to leverage both the understanding (encoder) and generation (decoder) strengths.

## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- **Benefits**:
  1) **Improved generalization**: Models learn general patterns of language that apply across many different topics.
  2) **Less labeled data**: Because the model already "knows" language, it only needs a few examples to learn a specific task.
  3) **Faster fine-tuning**: Adjusting a pre-trained model takes much less time than training one from scratch.
  4) **Transfer learning**: Knowledge gained from a large dataset (like Wikipedia) is successfully applied to specific domains (like legal or medical text).
  5) **Robustness**: Pre-trained models are generally better at handling variations and noise in input text.

- **Ethical concerns**: Models can amplify societal biases present in training data, generate convincing misinformation (hallucinations), be misused for phishing/spam, or leak private information found in the training set.

- **Mitigations**: Using curated datasets to reduce bias, applying differential privacy, implementing safety guardrails (RLHF), and conducting regular external audits to test for harmful outputs.

## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- **Mechanism**: The model creates Query (Q), Key (K), and Value (V) vectors for each word. It calculates a score by comparing Q and K, applies softmax to get weights, and uses these weights to sum the Values, allowing the model to focus on relevant words.
- **Multi-Head**: Using multiple heads allows the model to attend to information from different representation subspaces simultaneously, capturing diverse linguistic relationships.
- **Example**: "The bank was closed because of the river flood."
  - *Head 1* might link "bank" to "closed" (financial context).
  - *Head 2* might link "bank" to "river" (geographic context), helping the model disambiguate the word.

### Pre-training Objectives
- **MLM vs CLM**: MLM hides words in the middle of a sentence (bidirectional context), while CLM predicts the next word in a sequence (unidirectional context).
- **Preference**: Use MLM for understanding tasks (classification); use CLM for generative tasks (storytelling).
- **NSP**: Early BERT used Next Sentence Prediction to understand relationships between sentences, but modern models often find it unnecessary as long-context training provides similar signals more efficiently.

### Transformer Model Selection
- **Sentiment on reviews**: Encoder-only (BERT) - Best for understanding context and classifying text.
- **Conversational chatbot**: Decoder-only (GPT) - Optimized for generating fluent, creative responses.
- **Technical document translation**: Encoder-Decoder (T5/BART) - The encoder processes the source language while the decoder generates the target language.

### Positional Encoding
- **Purpose**: Transformers process all words in parallel, so they don't inherently know word order. Positional encoding adds a signal to each word embedding to indicate its location in the sequence.
- **Failure**: Without it, "The cat ate the mouse" and "The mouse ate the cat" would look identical to the model.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- **T=0.2**: The model becomes very focused and deterministic, picking only the most likely words (low diversity).
- **T=1.5**: The model becomes very random and "creative," often producing nonsensical or highly unexpected text.
- **T=1.0**: The default balance; the model follows its original learned probability distribution.

### 2) Application Design
- **Bedtime stories**: Use **T=0.7 to 0.9**. This allows for enough creativity to make the story interesting while maintaining enough coherence so the plot makes sense.
- **Financial reports**: Use **T=0 or 0.1**. Accuracy is critical here; you want the most likely, factual prediction with no "creative" hallucination.

### 3) Temperature & Bias
- Higher temperature can sometimes surface rarer, more diverse perspectives, but it can also trigger highly biased or offensive associations that were latent in the data. Conversely, very low temperature (T=0.1) often causes the model to default to the most common (and potentially stereotypical) majority bias found in its training set.

### (Optional) Quick Generation Demo Across Temperatures

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # TODO: choose a causal LM
prompt = "Artificial intelligence will"  # TODO: write your own prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Temperature:", T)
    print(out[0]["generated_text"])  # Inspect how style changes with T
```
