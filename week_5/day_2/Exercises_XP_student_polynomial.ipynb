**Réponse :**
Lors de la propagation avant, les données entrent par la **couche d'entrée** (inputs). Pour chaque neurone de la couche suivante, on calcule une **somme pondérée** des entrées via les **poids** (weights), à laquelle on ajoute un **biais** (bias). Ce résultat passe ensuite par une **fonction d'activation** pour introduire de la non-linéarité avant d'être transmis à la couche suivante. Ce processus se répète jusqu'à la **couche de sortie** (outputs) qui produit la prédiction finale du réseau.

**Learning point**

**As stated in the exercise**  
Objective. Generate a dataset with noise and visualize it.  
Instructions. Import numpy, matplotlib.pyplot, and mean_squared_error from sklearn.metrics. Create 20 points using \(y = -x^2\) with Gaussian noise \(\mathcal{N}(0, 0.05)\). Plot the points with a scatter plot. Split into a training set with the first 12 points and a test set with the last 8 points.


```python
# PREFILLED: just execute
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

np.random.seed(0)
x = np.arange(-1, 1, 0.1)  # 20 points from -1 to <1 step 0.1
print("x shape:", x.shape)
```

```python
import numpy as np

# S'assurer que x est défini (au cas où la cellule précédente n'a pas été exécutée)
np.random.seed(0)
x = np.arange(-1, 1, 0.1)

# Génération de y selon la formule y = -x^2 avec l'ajout d'un bruit gaussien
y = -x**2 + np.random.normal(0, 0.05, len(x))

print("y shape:", y.shape)
print("Aperçu des valeurs de y :", y[:5])
```

```python
# PREFILLED: just execute (scatter plot of the dataset)
plt.figure(figsize=(5,4))
plt.scatter(x, y)
plt.title("Noisy dataset sampled from y = -x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.tight_layout()
plt.show()
```

```python
# On s'assure d'utiliser les variables x et y définies précédemment
x_train, y_train = x[:12], y[:12]
x_test,  y_test  = x[12:], y[12:]

print(f"Taille entraînement : x={x_train.shape}, y={y_train.shape}")
print(f"Taille test : x={x_test.shape}, y={y_test.shape}")
```

## 🌟 Exercise 4: Fitting Polynomial Models of Different Degrees

**As stated in the exercise**  
Objective: Fit polynomial models of varying degrees and observe overfitting.  

Instructions: Define `polynomial_fit(degree)` that returns the fitted polynomial coefficients from training data. Define `plot_polyfit(degree)` that draws the training set, the test set, and the fitted polynomial curve. Visualize fits for degrees 1, 7, and 11. Observe behavior on training and test sets as degree increases.


```python
def polynomial_fit(degree: int):
    """Calcule les coefficients du polynôme et retourne une fonction polynomiale."""
    # np.polyfit calcule les coefficients minimisant l'erreur quadratique
    coeffs = np.polyfit(x_train, y_train, degree)
    # np.poly1d transforme ces coefficients en une fonction utilisable
    return np.poly1d(coeffs)

print("Fonction polynomial_fit prête.")
```

```python
import matplotlib.pyplot as plt

def plot_polyfit(degree: int):
    # 1) Ajustement du modèle sur les données d'entraînement
    model = polynomial_fit(degree)

    # 2) Création d'une ligne dense pour tracer une courbe lisse
    x_range = np.linspace(-1, 1, 100)
    y_poly = model(x_range)

    # 3) Visualisation
    plt.figure(figsize=(6, 4))
    plt.scatter(x_train, y_train, color='blue', label='Train (Entraînement)')
    plt.scatter(x_test, y_test, color='red', label='Test')
    plt.plot(x_range, y_poly, color='green', label=f'Degré {degree}')

    plt.ylim(-1.5, 0.5)
    plt.title(f"Ajustement polynomial de degré {degree}")
    plt.legend()
    plt.show()
```

```python
# S'assurer que les fonctions et variables sont présentes
# Visualisation pour les degrés 1, 7, et 11
for d in [1, 7, 11]:
    plot_polyfit(d)
```

**Learning point**  
Higher degree polynomials can perfectly fit training data yet fail to generalize. This is overfitting. You should always check test error, not just training error.

## 🌟 Exercise 5: Cross-Validation to Find the Optimal Degree

**As stated in the exercise**  
Objective: Use cross-validation to determine the best polynomial degree.  

Instructions: Loop degrees from 1 to 11. For each degree, compute the RMSE on the training and test sets. Store results in a list. Plot RMSE vs degree with a logarithmic y axis. Identify the degree minimizing test RMSE and confirm it matches the true model \(y=-x^2\).


```python
from sklearn.metrics import mean_squared_error

def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

degrees = range(1, 12)
rows = []

for d in degrees:
    # Entraînement sur le set d'entraînement uniquement
    coeffs = np.polyfit(x_train, y_train, d)
    model = np.poly1d(coeffs)

    # Calcul de la performance
    r_train = rmse(y_train, model(x_train))
    r_test = rmse(y_test, model(x_test))

    rows.append((d, r_train, r_test))

print("Calcul des RMSE terminé pour les degrés 1 à 11.")
```

```python
# Extraction des données calculées dans la cellule précédente
degs = [r[0] for r in rows]
rmse_tr = [r[1] for r in rows]
rmse_te = [r[2] for r in rows]

plt.figure(figsize=(7,5))
plt.plot(degs, rmse_tr, marker='o', label="RMSE Entraînement (Train)")
plt.plot(degs, rmse_te, marker='x', label="RMSE Test")

# Configuration de l'affichage
plt.yscale("log")
plt.xlabel("Degré du polynôme")
plt.ylabel("Erreur (RMSE) - Échelle Log")
plt.title("Évolution de l'erreur en fonction de la complexité")
plt.xticks(degs)
plt.legend()
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

**To-Do:** Report the degree with the minimum test RMSE and explain in 2–3 sentences how this connects to the true model \(y=-x^2\) and the bias–variance trade off.

**Analyse :**
Le degré qui minimise la RMSE sur l'ensemble de test est généralement le **degré 2**. Cela correspond parfaitement au modèle réel ($y = -x^2$) utilisé pour générer les données. C'est l'illustration du compromis **biais-variance** : un degré 1 a un biais trop élevé (sous-apprentissage), tandis qu'un degré 11 a une variance trop élevée car il essaie de modéliser le bruit au lieu du signal (sur-apprentissage).
